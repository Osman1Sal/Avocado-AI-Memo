```python
import pandas as pd  


avo = pd.concat([pd.read_csv('2025data.csv'), pd.read_csv('2026data.csv')])
avo['Year'] = pd.to_datetime(avo['Current Year Week Ending']).dt.year
d = avo[avo['Period'] <= 32].copy()                          # same 32 weeks in both years
d['Rev'] = d['ASP Current Year'] * d['Total Bulk and Bags']  # est. dollar sales (no revenue column in file)
us = d[d['Geography'] == 'Total U.S.']
regions = ['California','West','Plains','South Central','Southeast','Midsouth','Great Lakes','Northeast']

def yoy(data, by='Scope'):
    """Volume, est. $ sales and weighted ASP for 2025 vs 2026, with % change."""
    t = data.assign(Scope='Total U.S.').pivot_table(
        index=by, columns='Year', values=['Total Bulk and Bags', 'Rev'], aggfunc='sum') / 1e6
    parts = {'Vol (M)': t['Total Bulk and Bags'], 'Est $ (M)': t['Rev'],
             'ASP': t['Rev'] / t['Total Bulk and Bags']}
    out = pd.concat({k: v.assign(**{'%': (v[2026] / v[2025] - 1) * 100}) for k, v in parts.items()}, axis=1)
    return out.round(3)
```

```python
display(us.groupby('Year')['Current Year Week Ending'].agg(['min','max','nunique']))
yoy(us)
```

```python
yoy(us.assign(Block=(us['Period'] - 1) // 4 + 1), ['Block'])
```

```python
cols = ['4046','4225','4770','TotalBagged','Sml Bagged','Lrg Bagged','X-Lrg Bagged','Total Bulk and Bags']
mix = us.groupby('Year')[cols].sum().T
mix['% chg'] = (mix[2026] / mix[2025] - 1) * 100
mix[['share 2025','share 2026']] = (mix[[2025,2026]] / mix.loc['Total Bulk and Bags', [2025,2026]] * 100).values
mix.round(1)
```

```python
reg = d[d['Geography'].isin(regions)]
print('Regions sum to Total U.S.:', round(reg['Total Bulk and Bags'].sum()) == round(us['Total Bulk and Bags'].sum()))
yoy(reg, 'Geography').sort_values(('Vol (M)', '%'), ascending=False)
```
